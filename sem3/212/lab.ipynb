```python
def calc(teor, exp):
    return abs(teor - exp) / teor
```

```python
def calc_tau(t: np.ndarray,U: np.ndarray, A: float):
  y = np.log(1-U/A)
  k, b = np.polyfit(t, y, 1)
  return -1/k

def calc_Q(tau, T):
  return np.pi * tau / T

def print_tau(tau,L,R):
  tau_th = 2*L/R
  print(f"tau_exp: {tau*1000:.2f}mc")
  print(f"tau_th: {tau_th*1000:.2f}mc")
  print(f"error: {calc(tau_th,tau):.2f}")
  print()

Rl = 15

R1 = 10 + Rl
t1 = (np.array([1.5,2.7,4])-0.8) * 1e-3
T1 = np.mean(t1[1:] - t1[:-1])
U = np.array([9,12,12.9])
A = 13
tau1 = calc_tau(t1,U,A)
q1 = calc_Q(tau1, t1[1]-t1[0])

R2 = 20 + Rl
t2 = np.array([0.5,1.78,3]) * 1e-3
T2 = np.mean(t2[1:] - t2[:-1])
U = np.array([9,13.2,13.9])
A = 14
tau2 = calc_tau(t2,U,A)
q2 = calc_Q(tau2, t2[1]-t2[0])

R3 = 30 + Rl
t3 = np.array([0.38,1.42,2.65]) * 1e-3
T3 = np.mean(t3[1:] - t3[:-1])
U = np.array([4.5,14,14.9])
A = 15
tau3 = calc_tau(t3,U,A)
q3 = calc_Q(tau3, t3[1]-t3[0])

R = np.array([R1,R2,R3])
tau_exp = np.array([tau1,tau2,tau3])
tau_th = 2 * L / R
T = np.array([T1,T2,T3])
T = np.full(shape=(3,), fill_value=1/810)
# Q_th = np.pi * tau_th / T
Q_th = 1 / R * (L/C)**0.5
Q_exp = np.pi * tau_exp / T

df = pd.DataFrame({"R, Om" : R, "tau_exp, mc": tau_exp * 1000, "tau_th": tau_th * 1000, "tau error": calc(tau_th,tau_exp), "Q_exp": Q_exp, "Q_th": Q_th, "Q error": calc(Q_th,Q_exp)})
df
```
